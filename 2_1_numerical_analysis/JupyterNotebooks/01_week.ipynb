```python
def factorial(x):
    val=1
    for i in range(1,x+1):
        val=val*i
    return val

xfin=10
## Test factorial function
for i in range(1,xfin):
    print(f'{i}! = {factorial(i)}')

import math
print('---')
for i in range(1,xfin):
    print(math.factorial(i))
```

```python
def sinx_taylor(x,fin):
    val=0.
    for n in range(0,fin+1):
        term=(-1)**n/(factorial(2*n+1))*(x**(2*n+1))
        #print(f'{n}-th term: {term}')
        val+=term
    return val

x=3
n=3
error=sinx_taylor(x,n)-np.sin(x)
print('error:',error)
```

```python
import matplotlib.pyplot as plt
import numpy as np
xs=np.linspace(-1.5*np.pi,1.5*np.pi)

plt.plot(xs,np.sin(xs),label='actual')
for n in range(2,5):
    ys=sinx_taylor(xs,n)
    plt.plot(xs,ys,label=f'Taylor series n={n}')
plt.legend()
plt.xlim(-4,4)
plt.ylim(-2,2)
```
