## Numpy

```python
import numpy as np
print(np.__version__)
print(np.__file__)
arr = np.array([[1, 2, 3], [4, 5, 6]])
print(arr.shape)   # (2, 3) → 2행 3열
print(arr.ndim)    # 차원 수 → 2
print(arr.size)    # 전체 원소 개수 → 6
print(arr.dtype)   # 데이터 타입 → int64 (환경에 따라 다름)
print(arr.flatten())
print(arr.flatten().sum())
```

```python
arr.flatten().sum()
```

```python
print(arr)
#print(arr[1::,1::])
#print(arr[::2,::2])
print(arr[::-1,::-1])
print(arr[::-1,::])
print(arr[::,::-1])
```

```python
a=np.array([1,2,3])
b=np.array([4,5,6])
dotprod=0.
for i in range(len(a)):
    dotprod+=a[i]*b[i]
print(dotprod)
dotprod=(a*b).sum()
print(dotprod)
```
